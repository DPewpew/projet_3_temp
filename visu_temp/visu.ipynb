```python
def convert_estimated_owners_to_numeric(estimated_owners):
    if isinstance(estimated_owners, str):
        if " - " in estimated_owners: min_owners, max_owners = estimated_owners.split(" - ")
        else: min_owners = max_owners = estimated_owners
        try:
            return int((int(min_owners) + int(max_owners)) / 2)
        except ValueError:
            return 0
    else:
        return estimated_owners
```

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("../data/cleaned")

# === Load ===
games = pd.read_parquet(DATA_DIR / "games_clean.parquet")
langs = pd.read_parquet(DATA_DIR / "game_languages_clean.parquet")
cats = pd.read_parquet(DATA_DIR / "game_categories_clean.parquet")
genres = pd.read_parquet(DATA_DIR / "game_genres_clean.parquet")

print("=== SHAPES ===")
print("games:", games.shape)
print("languages:", langs.shape)
print("categories:", cats.shape)
print("genres:", genres.shape)


# =====================================================
# 1️⃣ GAMES CLEAN CHECK
# =====================================================

print("\n=== GAMES CHECK ===")

print("Unique appid:", games["appid"].nunique())
print("Total rows:", len(games))

print("\nNull % per column:")
display((games.isna().mean() * 100).round(2).sort_values(ascending=False))

print("\nRelease year range:")
print("Min:", games["release_year"].min())
print("Max:", games["release_year"].max())

print("\nEstimated owners numeric range:")
print("Min:", games["estimated_owners_numeric"].min())
print("Max:", games["estimated_owners_numeric"].max())


# =====================================================
# 2️⃣ LANGUAGES CHECK
# =====================================================

print("\n=== LANGUAGES CHECK ===")

print("Null rows:", langs.isna().sum().sum())
print("Duplicate (appid, language):",
      langs.duplicated(subset=["appid", "language"]).sum())

print("\nTop 20 languages:")
display(langs["language"].value_counts().head(20))


# =====================================================
# 3️⃣ CATEGORIES CHECK
# =====================================================

print("\n=== CATEGORIES CHECK ===")

print("Duplicate (appid, category):",
      cats.duplicated(subset=["appid", "category"]).sum())

print("\nTop 20 categories:")
display(cats["category"].value_counts().head(20))


# =====================================================
# 4️⃣ GENRES CHECK
# =====================================================

print("\n=== GENRES CHECK ===")

print("Duplicate (appid, genre):",
      genres.duplicated(subset=["appid", "genre"]).sum())

print("\nTop 20 genres:")
display(genres["genre"].value_counts().head(20))


# =====================================================
# 5️⃣ Integrity cross check
# =====================================================

print("\n=== CROSS CHECK APPID CONSISTENCY ===")

games_ids = set(games["appid"])

print("Languages missing appid in games:",
      (~langs["appid"].isin(games_ids)).sum())

print("Categories missing appid in games:",
      (~cats["appid"].isin(games_ids)).sum())

print("Genres missing appid in games:",
      (~genres["appid"].isin(games_ids)).sum())
```
